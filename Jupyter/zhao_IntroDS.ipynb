### Load Necessary Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Basic Graph

```python
x = [0,1,2,3,4]
y = [0,2,4,6,8]

# Resize graph
# figsize gives x-to-y ratio
# dpi specifies pixels per inch (300 is recommended for saving)
plt.figure(figsize=(8,5), dpi=100)

# -------- Line 1 --------

# Keyword argument notation
#plt.plot(x,y, label='2x', color='red', linewidth=2, marker='.', linestyle='--', markersize=10, markeredgecolor='blue')

# Shorthand notation (fmt stands for format)
# fmt = '[color][marker][linestyle]'
plt.plot(x,y, 'b^--', label='2x')

# -------- Line 2 -------- 

# Select interval to plot points
x2 = np.arange(0,4.5,0.5)

# Plot part of graph as line
plt.plot(x2[:6], x2[:6]**2, 'r', label='X^2')

# Plot remainder of graph dotted
plt.plot(x2[5:], x2[5:]**2, 'r--')

# ------------------------ 

# Add title (specify font parameters with fontdict)
plt.title('Our First Graph!', fontdict={'fontname': 'Comic Sans MS', 'fontsize': 20})

# X and Y labels
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

# X and Y tickmarks (scale of graph)
plt.xticks([0,1,2,3,4])
#plt.yticks([0,2,4,6,8,10])

# Add legend
plt.legend()

# Save figure (dpi=300 recommended for saving so graph has high resolution)
plt.savefig('mygraph.png', dpi=300)

# Show plot
plt.show()
```

### Bar Chart

```python
labels = ['A', 'B', 'C']
values = [1,4,2]

bars = plt.bar(labels, values)

# Set patterns
bars[0].set_hatch('/')
bars[1].set_hatch('O')
bars[2].set_hatch('*')

# More efficient way
patterns = ['/', 'O', '*']
for bar in bars:
    bar.set_hatch(patterns.pop(0))

plt.figure(figsize=(6,4))

plt.show()
```
