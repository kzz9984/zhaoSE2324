### Loading Data into Pandas

```python
import pandas as pd

df = pd.read_csv('pokemon_data.csv') # Data loaded in object type called dataframe

print(df.head(3)) # Default is 5
print(df.tail(3))

# Can load in data of various formats
df_xlsx = pd.read_excel('pokemon_data.xlsx')
df_txt = pd.read_csv('pokemon_data.txt', delimiter='\t') # File is tab-separated
```

### Reading Data in Pandas

```python
# Read headers
print(df.columns)

# Read specific columns
print(df['Name'][0:5])
print(df.Name[0:5]) # Only works for one-word column names

# Read multiple columns
print(df[['Name', 'Type 1', 'HP']])

# Read specific rows
print(df.iloc[1]) # iloc stands for integer location
print(df.iloc[0:4])

# Iterate through rows
for index, row in df[:20].iterrows():
    print(index, row['Name'])

# Find rows by content (instead of integer location)
print(df.loc[df['Type 1'] == "Fire"])

# Read a specific location (R,C)
print(df.iloc[2,1])
```

### Sorting/Describing Data

```python
# Gives column statistics (mean, std, etc.)
print(df.describe())

# Sort
print(df.sort_values('Name', ascending=False))
df.sort_values(['Type 1', 'HP'], ascending=[1,0]) # Ascending parameter set for each column
```

### Making Changes to Data

```python
# Simple way to do it
df['Total'] = df['HP'] + df['Attack'] + df['Defense'] + df['Sp. Atk'] + df['Sp. Def'] + df['Speed']

# Dropping a column
df = df.drop(columns=['Total'])

# More elegant way to do it
df['Total'] = df.iloc[:, 4:10].sum(axis=1)
# (However, it is safer to get indices by column name instead of hardcoding them)

# Reorder columns
cols = list(df.columns) # Get all columns as a list
df = df[cols[0:4] + [cols[-1]] + cols[4:12]] # Move final column to position 4
# "cols[-1]" is bracketed so it can be concatenated to the other lists

df.head(5)
```

### Saving Data (Exporting into Desired Format)

```python
df.to_csv('modified.csv', index=False) # index=False removes the index column

# Other formats
#df.to_excel('modified.xlsx', index=False)
#df.to_csv('modified.txt', index=False, sep='\t')
```

### Filtering Data

```python
# Passing in multiple conditions (parentheses are necessary)
# & = AND
# | = OR
new_df = df.loc[(df['Type 1'] == 'Grass') & (df['Type 2'] == 'Poison') & (df['HP'] > 70)]

# The new dataframe keeps the old indices, but can reset them
# drop=True removes column with old indices
# inplace=True avoids having to assign it to new_df
new_df.reset_index(drop=True, inplace=True) 

print(new_df)

#new_df.to_csv('filtered.csv')

# Drop all rows where Name contains 'Mega'
# ~ = NOT
print(df.loc[~df['Name'].str.contains('Mega')])

# Using regex to filter
import re

# Find rows where Type 1 is Fire or Grass
# flags=re.I ignores case in regex
print(df.loc[df['Type 1'].str.contains('fire|grass', flags=re.I, regex=True)])

# Find rows where Name starts with 'pi'
df.loc[df['Name'].str.contains('^pi[a-z]*', flags=re.I, regex=True)]
```

### Conditional Changes

```python
# Change all Type 1 'Fire' to 'Flamer' (and back)
df.loc[df['Type 1'] == 'Fire', 'Type 1'] = 'Flamer'
df.loc[df['Type 1'] == 'Flamer', 'Type 1'] = 'Fire'

# Make all Type 1 Fire rows legendary
df.loc[df['Type 1'] == 'Fire', 'Legendary'] = True

# Change multiple columns at a time (pass in list)
df.loc[df['Total'] > 500, ['Generation','Legendary']] = ['Test 1', 'Test 2']

# Load checkpoint to reset changes
df= pd.read_csv('modified.csv')

df
```

### Aggregate Statistics (Groupby)

```python
df = pd.read_csv('modified.csv')

# Find average stats grouped by Type 1, sorted by highest defense
# Note (this wasn't in the video, but from a comment):
#   Pandas v2.0.0 changed numeric_only default value to False
#   Setting it to True ensures it won't throw errors like "can not convert strings"
print(df.groupby(['Type 1']).mean(numeric_only=True).sort_values('Defense', ascending=False))

# Find sums of stats grouped by Type 1
print(df.groupby(['Type 1']).sum(numeric_only=True))

# Find counts of each Type 1 & 2 combination
df['count'] = 1 # Give each row a count column = 1
df.groupby(['Type 1', 'Type 2']).count()['count']
```

### Working with Large Amounts of Data

```python
# Create empty dataframe with same column names as original dataframe
new_df = pd.DataFrame(columns=df.columns)

# Load in data 5 rows at a time
for df in pd.read_csv('modified.csv', chunksize=5):
    results = df.groupby(['Type 1']).count()
    
    # Add on the new rows for this chunk
    new_df = pd.concat([new_df, results])

new_df
```
